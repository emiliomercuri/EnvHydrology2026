### Exemplo

Calcular $y$ para canal retangular com $B = 2$ m.

Dados
* $n = 0.015$  (s/[m$^{1/3}$])
* $S_o = 0.025$
* $Q = 9.26$ m$^3$/s

Sobre a fórmula de Manning:

https://en.wikipedia.org/wiki/Manning_formula

```python
Lc = 1000 # m
q0 = 0.00926 # m3/(s.m)
Q = Lc*q0
Q
```

```python
n = 0.015
S0 = 0.025
```

```python
# chute inicial
y = 1.0 # m
tol = 0.0001
```

```python
def vazao(y):
  n = 0.015
  S0 = 0.025
  return (1.0/n)*(S0**(0.5))*((2*y)**(5/3))/((2+2*y)**(2/3))
```

```python
def y_novo(y,Q):
  return y - (1-9.26/Q)*(y*(1+y))/(1.667+y)
```

```python
def erro(Q):
  return 9.26 - Q
```

```python
Q = vazao(1)
print(Q)
```

```python
e = erro(vazao(y))
```

```python
i = 0
while abs(e) > tol:
  y = y_novo(y,Q)
  Q = vazao(y)
  e = erro(Q)
  i += 1
  print(i, y,Q,e)
```

```python
print("A altura do escoamento é: ", y, "m")
```
