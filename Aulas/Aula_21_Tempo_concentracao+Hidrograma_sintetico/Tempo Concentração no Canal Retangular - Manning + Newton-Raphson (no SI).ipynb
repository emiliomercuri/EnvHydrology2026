# Exemplo - Tempo de concentração num canal retangular

Calcular $y$ para canal retangular com $B = 2$ m e $L = 1000$ m.

Dados
* $n = 0.02$  (s/[m$^{1/3}$])
* $S_o = 0.05$
* $q0 = 0.001$ m$^3$/(s.m)

Sobre a fórmula de Manning:

https://en.wikipedia.org/wiki/Manning_formula

A vazão no final do canal é:
* $Q = q0*L = 1.0$ m$^3$/s

A velocidade depende da área, que depende da altura. Vamos fazer uma solução iterativa (Newton-Raphson) para diversas alturas.

### Solução para cada 200 metros do canal

```python
import numpy as np
```

```python
# Valor constante
q0 = 0.001 # m3/(s.m)
```

```python
def vazao(y):
  n = 0.02
  S0 = 0.05
  return (1.0/n)*(S0**(0.5))*((2*y)**(5/3))/((2+2*y)**(2/3))
```

```python
def y_novo(y,Q,Qnovo):
  return y - (1-Q/Qnovo)*(y*(1+y))/(1.667+y)
```

```python
def erro(Q,Qnovo):
  return Q - Qnovo
```

```python
for Lc in range(200,1200,200):
  print(Lc)
```

```python
q0*Lc
```

```python
Q_acumulado = 0
y_acumulado = 0
v = [0]
for Lc in range(200,1200,200):
  Q = Lc*q0
  print("-------------------------------")
  print("Comprimento do canal: ", Lc, "m", "Escoamento Direto (lateral):",Q,r"m3/s")

  # chute inicial
  y = 0.5 # m
  tol = 0.0001

  Qnovo = vazao(y)
  e = erro(Q,Qnovo)

  i = 0
  while abs(e) > tol:
  #for i in range(10):
    y = y_novo(y,Q,Qnovo)
    Qnovo = vazao(y)
    e = erro(Q,Qnovo)
    i += 1
    print(i, y, Qnovo, e)

  Q_acumulado += Qnovo
  y_acumulado += y

  area = 2*y_acumulado
  velocidade = Q_acumulado/area
  print("A velocidade do escoamento é: ", velocidade, "m/s")
  v.append(velocidade)
```
