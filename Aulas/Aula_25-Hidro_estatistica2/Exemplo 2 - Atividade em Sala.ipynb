Considere a função densidade de probabilidade da curva gaussiana:

$$
f(x)=\frac{1}{\sqrt{2 \pi} \sigma} \exp \left[-\frac{(x-\mu)^2}{2 \sigma^2}\right]
$$

A variável normal padrão é $z$:
$$
z = \frac{x-\mu}{\sigma}
$$

A **distribuição normal padrão** tem função densidade de probabilidade:

$$
f ( z ) \,=\, \frac{1} {\sqrt{2 \pi}} \, \, e^{-z^{2} / 2} \qquad\quad-\infty\, \leq\, z \, \leq\, \infty
$$

A função distribuição de probabilidade é:

$$
F(z)=\int_{-\infty}^x \frac{1}{\sqrt{2 \pi} } e ^{-u^2 / 2 } d u
$$

$F(z)$ não tem forma analítica. Os valores de $F(z)$ podem ser obtidos pelo polinômio:

$$
B={\frac{1} {2}} [ 1+0. 1 9 6 8 5 4 \vert z \vert+0. 1 1 5 1 9 4 \vert z \vert^{2}+0. 0 0 0 3 4 4 \vert z \vert^{3}+0. 0 1 9 5 2 7 \vert z \vert^{4} ]^{-4}
$$

onde $\vert z \vert$￼é o valor absoluto de $z$ e a distribuição normal padrão é:

$$
\begin{align} {{{F ( z )=B \qquad\mathrm{para} \ z < 0}}} \\ {{{=1-B \ \mathrm{para} \ z \geq0}}} \\ \end{align}
$$



### Questão da atividade em sala

Qual é a probabilidade de que a variável aleatória normal padrão z seja:
* a) Menor que -2?
* b) Menor que 1?
* c) Quanto vale $P(-2<z<1)$?

$$
B={\frac{1} {2}} [ 1+0. 1 9 6 8 5 4 \vert z \vert+0. 1 1 5 1 9 4 \vert z \vert^{2}+0. 0 0 0 3 4 4 \vert z \vert^{3}+0. 0 1 9 5 2 7 \vert z \vert^{4} ]^{-4}
$$

```python
def B(z):
  """
  B is a polynomial approximation of the cumulative distribution function of the standard normal distribution.
  """
  return (1 / 2) * (1 + 0.196854 * abs(z) + 0.115194 * abs(z)**2 + 0.000344 * abs(z)**3 + 0.019527 * abs(z)**4)**(-4)

def normal_cdf(z):
  """
  Cumulative distribution function of the standard normal distribution.
  """
  if z < 0:
    return B(z)
  else:
    return 1 - B(z)
```

```python
# a)
print("P(z < -2) = F(-2) = ", normal_cdf(-2))

# b)
print("P(z < 1) = F(1) =", normal_cdf(1))

# c)
print("P(-2 < z < -1) = F(1) - F(-2) =", normal_cdf(1) - normal_cdf(-2))
```
